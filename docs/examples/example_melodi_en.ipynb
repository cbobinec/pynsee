## VI. Using the polars module, which is similar to dplyr

```python
# Import the polars module installed via the command "pip install polars" in the VS Code terminal
# pl is an alias for the polars module, which is a data manipulation library in Python. It provides data structures and functions to work with data frames, similar to pandas, but with improved performance for large datasets.
import polars as pl
```

### VI.a Import a dataset

```python
# Extract the dataset "DS_RP_MENAGES_PRINC_IRIS" selecting the measure "POP" or "ONEPERS", civil status "_T", and age "Y_GE15" using the get_data function of the melodi submodule
POP_IRIS=pynsee.get_data("https://api.insee.fr/melodi/data/DS_RP_MENAGES_PRINC_IRIS?RP_MEASURE=POP&RP_MEASURE=ONEPERS&CIVIL_STATUS=_T&AGE=Y_GE15")
```

### VI.b Filter data

```python
# Convert the pandas dataframe into a polars dataframe, mandatory to use polars methods/functions
POP_IRIS=pl.from_pandas(POP_IRIS)

# Filter the POP_IRIS table to keep only the total population (RP_MEASURE=POP)
POP_IRIS=POP_IRIS.filter(pl.col("RP_MEASURE")=="POP")

# Some polars dataframe methods

## First and last 5 rows of the POP_IRIS dataframe
POP_IRIS.head(5)
POP_IRIS.tail(5)

## Number of rows and columns of the POP_IRIS dataframe
POP_IRIS.shape

## Display the type of each column of the POP_IRIS dataframe
POP_IRIS.schema

## Statistical description of the columns of the POP_IRIS dataframe
POP_IRIS.describe()
```

### VI.c Select columns and rename a variable

```python
# Select the columns "TIME_PERIOD", "GEO_OBJECT", "GEO", "LAB_IRIS", and "OBS_VALUE" from the POP_IRIS dataframe and rename "OBS_VALUE" to "POP_IRIS"
POP_IRIS=POP_IRIS.select("TIME_PERIOD","GEO_OBJECT","GEO","LAB_IRIS","OBS_VALUE").rename({"OBS_VALUE":"POP_IRIS"})  

# First 5 rows of the POP_IRIS dataframe
POP_IRIS.head(5)
```

### VI.d Merge tables

```python
# Import data to get the number of people with a Bac+5 or higher degree per IRIS
DIPLO_BAC_PLUS_5_OU_PLUS=pynsee.get_data("https://api.insee.fr/melodi/data/DS_RP_DIPLOMES_PRINC_IRIS?SEX=_T&EDUC=700_RP")

# Transform the pandas dataframe into a polars dataframe to use polars methods/functions
DIPLO_BAC_PLUS_5_OU_PLUS=pl.from_pandas(DIPLO_BAC_PLUS_5_OU_PLUS)

# Select only the relevant columns in the DIPLO_BAC_PLUS_5_OU_PLUS dataframe
DIPLO_BAC_PLUS_5_OU_PLUS=DIPLO_BAC_PLUS_5_OU_PLUS.select("TIME_PERIOD","GEO_OBJECT","GEO","LAB_IRIS","OBS_VALUE").rename({"OBS_VALUE":"POP_DIPLO_BAC_PLUS_5_OU_PLUS"})

# Merge the two dataframes POP_IRIS and DIPLO_BAC_PLUS_5_OU_PLUS on TIME_PERIOD, GEO_OBJECT, GEO, and LAB_IRIS columns
POP_IRIS_DIPLO_BAC_PLUS_5_OU_PLUS=POP_IRIS.join(DIPLO_BAC_PLUS_5_OU_PLUS, on=["TIME_PERIOD","GEO_OBJECT","GEO","LAB_IRIS"], how="inner")

# First 5 rows of the POP_IRIS_DIPLO_BAC_PLUS_5_OU_PLUS dataframe
POP_IRIS_DIPLO_BAC_PLUS_5_OU_PLUS.head(5)
```

### VI.e Create/modify a column

```python
# Calculate the percentage of people with a Bac+5 or higher degree per IRIS and round the result to 2 decimal places
POP_IRIS_DIPLO_BAC_PLUS_5_OU_PLUS = (
    POP_IRIS_DIPLO_BAC_PLUS_5_OU_PLUS
    .filter(pl.col("POP_IRIS") != 0) # Filter out rows where total population (POP_IRIS) is zero to prevent division by zero
    .with_columns(
        (
            pl.col("POP_DIPLO_BAC_PLUS_5_OU_PLUS")
            / pl.col("POP_IRIS")
            * 100 # Calculate the percentage of people with a Bac+5 or higher degree per IRIS
        )
        .round(2) # Round the result to 2 decimal places
        .alias("POURCENTAGE_DIPLO_BAC_PLUS_5_OU_PLUS") # Name the new column "POURCENTAGE_DIPLO_BAC_PLUS_5_OU_PLUS"
    )
)

# First 5 rows of the POP_IRIS_DIPLO_BAC_PLUS_5_OU_PLUS dataframe
POP_IRIS_DIPLO_BAC_PLUS_5_OU_PLUS.head(5)
```

### VI.f Calculate indicators by group, calculate the share of people per department with a Bac+5 or higher


```python
# Calculate the percentage of people with a Bac+5 or higher degree per department and round the result to 2 decimal places
POP_IRIS_DIPLO_BAC_PLUS_5_OU_PLUS_DEP = (
    POP_IRIS_DIPLO_BAC_PLUS_5_OU_PLUS
    .with_columns(
        # Create a new "DEP" column containing the department code from the "GEO" column
        DEP=pl.when(
            pl.col("GEO").str.starts_with("97") # For overseas departments (e.g., 971, 972, 973, 974, 976)
            | pl.col("GEO").str.starts_with("98")
        )
        .then(
            pl.col("GEO").str.slice(0, 3) # 3 first characters for overseas territories (e.g., 971, 976, ...)
        )  
        .otherwise(
            pl.col("GEO").str.slice(0, 2) # 2 first characters for mainland France (e.g., 75, 69)
        )
    )
    .group_by("DEP") # Group by department
    .agg(POP_DEP=pl.col("POP_IRIS").sum(), # Sum population per department, creating POP_DEP column
         DIPLO_DEP=pl.col("POP_DIPLO_BAC_PLUS_5_OU_PLUS").sum()) # Sum bac + 5 or higher per department, creating DIPLO_DEP column

    .with_columns(
        (
            pl.col("DIPLO_DEP")
            / pl.col("POP_DEP")
            * 100 # Calculate the percentage of people with a Bac+5 or higher degree per department
        )
        .round(2) # Round the result to 2 decimal places
        .alias("POURCENTAGE_DIPLO_BAC_PLUS_5_OU_PLUS_DEP") # Name the new column "POURCENTAGE_DIPLO_BAC_PLUS_5_OU_PLUS_DEP"
    )
    .sort(by=["POURCENTAGE_DIPLO_BAC_PLUS_5_OU_PLUS_DEP"], descending=[True]))# Sort in descending order by the percentage of people with a Bac+5 or higher degree per department

# 5 first rows of the dataframe POP_IRIS_DIPLO_BAC_PLUS_5_OU_PLUS_DEP
POP_IRIS_DIPLO_BAC_PLUS_5_OU_PLUS_DEP.head(5)
```

### VI.g Sort a table

```python
# Sort POP_IRIS_DIPLO_BAC_PLUS_5_OU_PLUS by descending percentage of people with a Bac+5 or higher degree
POP_IRIS_DIPLO_BAC_PLUS_5_OU_PLUS = POP_IRIS_DIPLO_BAC_PLUS_5_OU_PLUS.sort(by=["POURCENTAGE_DIPLO_BAC_PLUS_5_OU_PLUS"], descending=[True])

# 5 first rows of the dataframe POP_IRIS_DIPLO_BAC_PLUS_5_OU_PLUS
POP_IRIS_DIPLO_BAC_PLUS_5_OU_PLUS.head(5)
```
